# HOW TO TRAIN OF THE MODEL

```python
x = dataset[["discounted_price"]]
y = dataset["actual_price"]
```

```python
x.ndim
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
lr.fit(x_train, y_train)
```

```python
plt.scatter(dataset["discounted_price"],dataset["actual_price"])
plt.title("LinearRegression: Discounted_price - Actual_price", pad=25)
plt.plot(x_train,lr.predict(x_train), color="red")
plt.xlabel("discounted_price")
plt.ylabel("actual_price")
plt.show()
```

```python
lr.score(x_test, y_test)*100
```

# MEA (MEAN ABSOLUTE ERROR)

```python
print(mean_absolute_error(y_test,lr.predict(x_test)))
```

# MSE (MEAN SQUARED ERROR)

```python
print(mean_squared_error(y_test,lr.predict(x_test)))
```

# RMSE (ROOT MEAN SQUARED ERROR)

```python
print(np.sqrt(mean_squared_error(y_test,lr.predict(x_test))))
```
